```python
import os
os.environ["OPENROUTER_API_KEY"] = "YOUR_OPENROUTER_API_KEY"
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.checkpoint.memory import MemorySaver
from langgraph.types import interrupt, Command
from langchain_openai import ChatOpenAI
```

```python
class EmailState(TypedDict):              #state with 5 field
    topic        : str                    #as we keep going thriugh nodes..we update these fields..one node can also update multiple fields
    draft        : str
    approved     : bool
    final_status : str
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o-mini",
    api_key=os.environ["OPENROUTER_API_KEY"],
    base_url="https://openrouter.ai/api/v1",
    temperature=0.7
)
```

```python
def write_email(state: EmailState):
  print("writing the mail")

  response= llm.invoke(                                            #llm is called here
      f"write a short email about: {state['topic']}"               #these are the prompts for llm
      f"keep it under 100 words."
  )

  return {"draft": response.content}                               #saves this in draft field
```

```python
def human_review(state: EmailState):
    print("\n Email draft is ready.")
    print("\n EMAIL DRAFT")
    print(state["draft"])
    print("-------------------\n")

    decision = interrupt({                                           #the workflow freezes here
        "message" : "Do you want to send this email?",               #with a payload
        "draft"   : state["draft"]
    })

    return {"approved": decision == "yes"}                           #internally waits for human to say yes or no...then it comapres decision and send email works then
```

```python
def send_email(state: EmailState):
    if state["approved"]:
        print("Email Sent!")
        return {"final_status": "sent"}
    else:
        print("Email Cancelled.")
        return {"final_status": "cancelled"}
```

```python
builder = StateGraph(EmailState)                     #builder is a stategraph object like a blank canvas

builder.add_node("write_email",   write_email)
builder.add_node("human_review",  human_review)
builder.add_node("send_email",    send_email)

builder.add_edge(START,           "write_email")
builder.add_edge("write_email",   "human_review")
builder.add_edge("human_review",  "send_email")
builder.add_edge("send_email",     END)

checkpointer = MemorySaver()#RAM storage                          #data till checkpoint gets saved in memory saver
graph = builder.compile(checkpointer=checkpointer)          #locks the graph and turns into runnable machine
```

```python
config = {"configurable": {"thread_id": "email-001"}}

result = graph.invoke(                               #this starts the workflow from the start
    {"topic": "asking for 1 day leave"},
    config=config
)

print("\n⏸️  Graph is paused. Human review needed.")
print("\nInterrupt payload:")
print(result["__interrupt__"][0].value) #prints the payload data..0 is index that we have this data at 0th index
```

```python
human_decision = "yes"   # change to "no" to cancel

final = graph.invoke(                                    #LangGraph finds the frozen checkpoint in MemorySaver using thread_id resumes human_review right after interrupt()
    Command(resume=human_decision),
    config=config                        #same threadid to find the checkpoint
)

print("\nFinal status:", final["final_status"])
```
